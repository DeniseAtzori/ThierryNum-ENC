# Script for merging Yolo inferences (tag_1) in CSV file

```python
import pandas as pd

# Input files
report_file = "report_inference.csv"
images_file = "images_inference.csv"
output_file = "images_inference_ok.csv"

# Read CSV
df_report = pd.read_csv(report_file, sep=",")
df_images = pd.read_csv(images_file, sep=",")

# Conversion map
class_mapping = {
    "11 Photographie": "1.1. Photographie",
    "12 Reproduction de plan architectural": "1.2. Reproduction de plan architectural",
    "13 Autre document reproduit gravure dessin etc": "1.3. Autre document reproduit (gravure, dessin, etc)",
    "14 Matériel de conditionnement": "1.4. Matériel de conditionnement",
    "15 Autre": "1.5. Autre"
}

# Conversion
df_report["tag_1"] = df_report["Classe_Predetta"].map(class_mapping)

# Segnalation of wrong mapping
missing = df_report[df_report["tag_1"].isna()]
if not missing.empty:
    print("⚠️ Attenzione: alcune classi non sono state mappate:")
    print(missing["Classe_Predetta"].unique())

# Dataframe creation
df_new_rows = pd.DataFrame({
    "title": df_report["Filename"],
    "image_url": df_report["Path"],   # 👈 ora viene dal CSV di report
    "tag_1": df_report["tag_1"],
    "tag_2": ""                       # vuoto
})

# Merge
df_final = pd.concat([df_images, df_new_rows], ignore_index=True)

# Save
df_final.to_csv(output_file, index=False)

print(f"✅ File generato: {output_file}")
```
