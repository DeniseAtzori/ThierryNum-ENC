# Script pour fusioner les resultats du clustering au fichier images

```python
import pandas as pd

# 1. Chargement des fichiers CSV
# Nous chargeons les deux fichiers dans des structures de données appelées DataFrames
df_clustering = pd.read_csv('clustering_results_efficientnet.csv')
df_inference = pd.read_csv('images_inference.csv')

# 2. Fusion des fichiers (Merge)
# Nous utilisons une jointure de type 'outer' (externe) pour gérer les trois cas :
# - 'on=['title', 'image_url']' : on utilise ces deux colonnes comme clés communes.
# - 'how='outer'' : cela garde toutes les lignes des deux fichiers.
# 
# Cette opération va :
# - Remplir 'clustering_id' pour les titres présents dans les deux fichiers.
# - Garder les lignes de 'images_inference' intactes si le titre n'est pas dans l'autre fichier.
# - Ajouter les nouvelles lignes de 'clustering_results' si le titre n'existait pas encore.
df_final = pd.merge(
    df_inference, 
    df_clustering, 
    on=['title', 'image_url'], 
    how='outer'
)

# 3. Réorganisation optionnelle (facultatif)
# Si tu souhaites que les colonnes soient dans un ordre spécifique, 
# tu peux les réorganiser ici.
cols = ['title', 'image_url', 'clustering_id', 'tag_1', 'tag_2', 'tag_3', 'tag_4', 'tag_5', 'tag_6', 'tag_7', 'tag_8', 'tag_9']
df_final = df_final[cols]

# 4. Sauvegarde du résultat
# On enregistre le résultat dans un nouveau fichier ou on écrase 'images_inference.csv'
df_final.to_csv('images_inference_updated.csv', index=False)

print("Le merge est terminé avec succès ! Le fichier 'images_inference_updated.csv' a été créé.")
```
